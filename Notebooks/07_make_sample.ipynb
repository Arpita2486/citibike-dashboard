```python
from pathlib import Path
import pandas as pd
import numpy as np

# Paths
PROJECT = Path.cwd()
if not (PROJECT / "Output").exists():
    PROJECT = PROJECT.parent if (PROJECT.parent / "Output").exists() else PROJECT
OUT = PROJECT / "Output"

full_path   = OUT / "citibike_weather_2022.csv.gz"                # if you saved full merged
fallback    = OUT / "citibike_weather_2022_sample_100k.csv"       # from Exercise 2.2
sample_path = OUT / "citibike_dashboard_sample.csv"

# Load full if available, else fallback
if full_path.exists():
    print("Loading full merged file:", full_path.name)
    df = pd.read_csv(full_path, low_memory=False)
else:
    print("Full file not found; loading fallback:", fallback.name)
    df = pd.read_csv(fallback, low_memory=False)

# Clean/types
df["date"] = pd.to_datetime(df["date"], errors="coerce").dt.date

# Random sample (seed=32) to keep under 25MB
n = min(len(df), 75_000)   # 75k rows is typically <25MB for these columns
df_s = df.sample(n=n, random_state=32).reset_index(drop=True)

df_s.to_csv(sample_path, index=False)
print("Saved sample:", sample_path, "rows:", len(df_s))
```
